```python
import pysam
import pandas as pd
```

```python


def bedpe_to_vcf(df, vcf_file, reference_genome):

    # Create a VCF file with pysam
    with pysam.VariantFile(vcf_file, 'w', header=create_vcf_header(reference_genome)) as vcf_out:
        for index, row in df.iterrows():
            # Create VCF record for each BEDPE entry
            vcf_record = create_vcf_record(row, vcf_out.header)
            vcf_out.write(vcf_record)

def create_vcf_header(reference_genome):
    header = pysam.VariantHeader()
    header.add_line("##fileformat=VCFv4.2")
    header.add_line(f"##reference={reference_genome}")
    header.add_line("##INFO=<ID=SVTYPE,Number=1,Type=String,Description=\"Type of structural variant\">")
    header.add_line("##INFO=<ID=SVLEN,Number=1,Type=Integer,Description=\"Length of structural variant\">")
    header.add_line("##INFO=<ID=END,Number=1,Type=Integer,Description=\"End position of the variant\">")
    header.add_line("##INFO=<ID=STRANDS,Number=1,Type=String,Description=\"Strand orientation of the variant\">")
    header.add_line("##INFO=<ID=REPLICATES,Number=1,Type=Integer,Description=\"Number of replicates called\">")
    header.add_sample("SAMPLE")

    return header

def create_vcf_record(row, header):
    chrom = row['Chrom1']
    pos = int(row['Pos1'])
    id = row['SV_id']
    ref = 'N'
    alt = f"<{row['SV_type']}>"
    qual = '.'
    filter = 'PASS'

    if row[['SV_type']] == "INS":
        end = int(row['Pos1']) + 1
    elif row[['SV_type']] == "DEL":
        end = int(row['Pos1']) + int(row['SV_Size_or_breakpoints_distance'])
    elif row[['SV_type']] == "DUP":
        end = int(row['Pos1']) + 1
    elif row[['SV_type']] == "TRA":
    
    info = {
        'SVTYPE': row['SV_type'],
        'SVLEN': row['SV_Size_or_breakpoints_distance'],
        'END': row['Pos2'],
        'STRANDS': row['SV_id'].split('_')[-1],
        'REPLICATES': row['Replicates_called']
    }

    record = header.new_record(
        contig=chrom,
        start=pos,
        id=id,
        alleles=(ref, alt),
        qual=qual,
        filter=filter,
        info=info
    )

    return record

```

```python
# Example usage
bedpe_file = '/Users/w620-admin/Desktop/high_conf_sv/high_conf_sv.tab'
vcf_file = 'output.vcf'
reference_genome = '/Users/w620-admin/Desktop/high_conf_sv/hg38.fa'

bedpe_to_vcf(bedpe_file, vcf_file, reference_genome)
```

```python
import csv

input_file = "/Users/w620-admin/Desktop/nf-core/dataset/SEQC2_SV_high_conf/13059_2022_2816_MOESM4_ESM.bedpe"
output_file = "/Users/w620-admin/Desktop/nf-core/dataset/SEQC2_SV_high_conf/13059_2022_2816_MOESM4_ESM.sv_transformed.vcf"

vcf_header = """##fileformat=VCFv4.2
##INFO=<ID=SVTYPE,Number=1,Type=String,Description="Type of structural variant">
##INFO=<ID=END,Number=1,Type=Integer,Description="End position of the variant">
##INFO=<ID=SVLEN,Number=1,Type=Integer,Description="Length of structural variant">
##FORMAT=<ID=GT,Number=1,Type=String,Description="Genotype">
#CHROM\tPOS\tID\tREF\tALT\tQUAL\tFILTER\tINFO\tFORMAT\tSAMPLE
"""

with open(input_file, "r") as infile, open(output_file, "w") as outfile:
    outfile.write(vcf_header)
    reader = csv.DictReader(infile, delimiter="\t")

    for row in reader:
        chrom1 = row['chrom1']
        start1 = int(row['start1'])
        chrom2 = row['chrom2']
        start2 = int(row['start2'])
        svtype = row['name']
        id_ = row['id']
        strand1 = row['strand1']
        strand2 = row['strand2']
        svlen = row['SV_Size_or_breakpoints_distance']

        ref = "N"
        qual = "."
        fltr = "PASS"
        fmt = "GT"
        sample = "1/1"

        if svtype in ["DEL", "DUP"]:
            end = start2
            info = f"SVTYPE={svtype};END={end};SVLEN={svlen}"
            alt = f"<{svtype}>"
        elif svtype == "INS":
            end = start1 + 1
            info = f"SVTYPE=INS;END={end};SVLEN={svlen}"
            alt = "<INS>"
        elif svtype == "TRA":
            # Breakend ALT formatting depending on strand orientation
            if strand1 == "+" and strand2 == "+":
                alt = f"N[{chrom2}:{start2}["
            elif strand1 == "+" and strand2 == "-":
                alt = f"N]{chrom2}:{start2}]"
            elif strand1 == "-" and strand2 == "+":
                alt = f"]{chrom2}:{start2}]N"
            elif strand1 == "-" and strand2 == "-":
                alt = f"[{chrom2}:{start2}[N"
            info = f"SVTYPE=BND"
        else:
            continue  # skip unknown SV types

        vcf_line = f"{chrom1}\t{start1}\t{id_}\t{ref}\t{alt}\t{qual}\t{fltr}\t{info}\t{fmt}\t{sample}\n"
        outfile.write(vcf_line)

print(f"✔ VCF successfully written to {output_file}")
```
